```python
import tensorflow as tf
dataset_path = "../data/dataset_farsi"
raw_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    image_size=(10,10),
    batch_size=32,
    color_mode="grayscale"
)
```

```python
import matplotlib.pyplot as plt

for images, labels in raw_ds.take(1):
    image = images[0]
    label = labels[0]

    plt.imshow(image.numpy().squeeze())
    plt.title(f"Label: {label.numpy()}")
    plt.axis("off")
    plt.show()
```

```python
print(image.numpy().shape)  # Should print (28, 28, 1)
```

## Dataset Summary

- classes: 10 digits (0-9)

- Image size: 28x28 grayscale

- Total images: 80000

- Format : PNG

- Labels: folder-based (0-9)

## Preprocessing Tests

```python
normalization_layer = tf.keras.layers.Rescaling(1./255)

train_ds = raw_ds.map(
    lambda x,y: (normalization_layer(x), y)
)
```

```python
for images, labels in train_ds.take(200):
    image = images[0]
    label = labels[0]

    plt.imshow(image.numpy().squeeze(), cmap="gray")
    plt.title(f"Label: {label.numpy()}")
    plt.axis("off")
    plt.show()
print(image.numpy)
```
