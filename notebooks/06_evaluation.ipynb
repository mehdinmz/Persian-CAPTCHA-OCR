# Model Evaluation & Confusion Matrix Analysis
This notebook evaluates the handwritten-digit model, computes accuracy, and visualizes confusion matrices.

```python
import sys
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
from pathlib import Path

# Ensure root is in path
sys.path.insert(0, os.path.abspath('..'))

from src.predictor import model
from scripts.evaluate import _get_sample_dir

print('Libraries and model loaded.')
```

## 1. Run Evaluation

```python
def run_eval(n_per_class=50):
    data_dir = _get_sample_dir()
    correct = 0
    total = 0
    all_preds = []
    all_labels = []
    
    for d in range(10):
        files = list((data_dir / str(d)).glob('*'))[:n_per_class]
        for f in files:
            img = cv2.imread(str(f), cv2.IMREAD_GRAYSCALE)
            c28 = cv2.resize(img, (28, 28)).astype(np.float32) / 255.0
            p = model.predict(c28.reshape(1, 28, 28, 1), verbose=0)[0]
            pred = int(np.argmax(p))
            all_preds.append(pred)
            all_labels.append(d)
            if pred == d: correct += 1
            total += 1
    
    print(f'Overall Accuracy: {correct/total:.1%}')
    return all_labels, all_preds

labels, preds = run_eval()
```

## 2. Confusion Matrix Visualization

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

cm = confusion_matrix(labels, preds)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()
```
